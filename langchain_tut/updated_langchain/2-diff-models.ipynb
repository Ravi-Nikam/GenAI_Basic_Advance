### Striming and batch 
#### striming means output generate little by little insted of full respose
 

```python
# model.stream("tell me  10 lines paragraph about Ai")
for chunk in model.stream("tell me  10 lines paragraph about Ai"):
    print(chunk.text,end= "|", flush=True)
```

#### Batching means sending multiple inputs together and processing them as a group

```python
responses=model.batch(
    ["why parrots have differance colur",
    "What is Machine Learning",
    "Why AI is important "]
)
# above all input goes once for processing (parellely it will go one by one for processing ) and generate the group output
for response in responses:
    print(response)
```

```python
model.batch(
    ["why parrots have differance colur",
    "What is Machine Learning",
    "Why AI is important "],
    config={
        "max_concurrency": 5, # how many parellel call you can made
        }
)

# if we give 10 question then it will give 5 5 to the LLM and generate the output
```
