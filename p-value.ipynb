```python
import numpy as np
from scipy import stats

# Generate data for Group A
# Mean = 75, Standard Deviation = 10, Number of values = 30
group_a = np.random.normal(75, 10, 30)

# Generate data for Group B
# Mean = 80, Standard Deviation = 12, Number of values = 30
group_b = np.random.normal(80, 12, 30)

# Perform Independent T-test
# It checks whether the mean of Group A and Group B
# are significantly different
t_statistics, p_value = stats.ttest_ind(group_a, group_b)

# Print the T-test statistic
print("t_statistics:", t_statistics)

# Print the p-value
print("p_value:", p_value)

# Set significance level
# Usually we use 0.05 (5%)
alpha = 0.05

# Compare p-value with significance level
if p_value <= alpha:
    # p-value is small → strong evidence against H0
    # Therefore, reject the null hypothesis
    print("Reject H0")
else:
    # p-value is large → not enough evidence against H0
    # Therefore, fail to reject the null hypothesis
    print("Fail to reject H0")
```
