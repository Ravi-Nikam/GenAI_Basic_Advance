```python
import pyttsx3
import speech_recognition as sr
import wikipedia

engine = pyttsx3.init()
engine.setProperty("rate", 175)


def speak(text):
    engine.say(text)
    engine.runAndWait()


def voice_wikipedia_search():
    recognizer = sr.Recognizer()

    with sr.Microphone() as source:
        recognizer.adjust_for_ambient_noise(source, duration=1)
        speak("What would you like to look up on Wikipedia?")

        try:
            audio = recognizer.listen(source, timeout=5, phrase_time_limit=5)
            query = recognizer.recognize_google(audio)

            wikipedia.set_lang("en")
            summary = wikipedia.summary(query, sentences=3)

            print(f"\nResult for '{query}':\n{summary}")
            speak(summary)

        except wikipedia.exceptions.DisambiguationError as e:
            fallback = e.options[0]
            summary = wikipedia.summary(fallback, sentences=3)
            print(f"\nResult for '{fallback}':\n{summary}")
            speak(summary)

        except Exception:
            speak("An error occurred.")


if __name__ == "__main__":
    voice_wikipedia_search()
```
