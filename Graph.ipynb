```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig,axs=plt.subplots(3,2,figsize=(14,12))
fig.suptitle("Graph Distribution")

data = np.random.normal(loc=0, scale=1, size=1000) #mean 0 sd =1
sns.histplot(data,kde=True,ax=axs[0,0],color="blue",stat='density',bins=30)
axs[0,0].set_title("Normal Distribution")
axs[0,0].grid(True)


```
