```python
from scipy import stats
import numpy as np
```

```python
data = [10,20,30,45,12,10,20,30]

x=stats.mode(data,keepdims=False)
print(x)
```

```python
# variance using numpy

data = [10,23,43,12,43]

population =np.var(data,ddof=0) # population variance
print(population)


population =np.var(data,ddof=1) # sample variance 1
print(population)
```

```python
import pandas as pd

seris=pd.Series(data)
print(seris)


sample_variance=seris.var()

popu_variance=seris.var(ddof=1)

print(sample_variance)

print(popu_variance)
```

```python
#using scipy

sample=stats.tvar(data)

population=stats.tvar(data,ddof=1)


print(sample)
print(population)
```

```python
# standard deviation using numpy
# ddof=0 → Population Standard Deviation
```

```python
sample_Std = np.std(data,ddof=1)
print("sample stadard deviation is ",sample_Std)

sample_Std = np.std(data,ddof=0)
print("population stadard deviation is ",sample_Std)
```

```python
# using pandas 

import pandas as pd

seris=pd.Series(data)
print(seris)


sample_variance=seris.std()

popu_variance=seris.std(ddof=0)

print(sample_variance)

print(popu_variance)
```

```python
import math

std=math.sqrt(stats.tvar(data))


std_popu=math.sqrt(stats.tvar(data,ddof=0))


print(std,std_popu)
```
